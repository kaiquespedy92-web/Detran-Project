```python
import pandas as pd
import os
```

```python
# Caminho da pasta
pasta = "/content/drive/MyDrive/Colab Notebooks/spark/data"
```

```python
meses = [
    "05-2026",
    "06-2026",
    "07-2026"
]
```

```python
def consolidar(tipo):
    tabelas = []
    for mes in meses:
        arquivo = os.path.join(
            pasta,
            f"{tipo}_{mes}.csv"
        )
        print(f"Lendo: {arquivo}")
        df = pd.read_csv(
            arquivo,
            sep=";",
            encoding="latin1" # Changed encoding from utf-8 to latin1
        )
        tabelas.append(df)

    # Consolidate after the loop finishes reading all files
    consolidado = pd.concat(
        tabelas,
        ignore_index=True
    )

    arquivo_saida = os.path.join(
        pasta,
        f"{tipo}_consolidado.csv"
    )

    consolidado.to_csv(
        arquivo_saida,
        sep=";",
        encoding="utf-8-sig", # Keep utf-8-sig for output to ensure compatibility with Excel
        index=False
    )
    print(f"\nArquivo criado: {arquivo_saida}")
    print(f"Linhas: {len(consolidado):,}")
    print(f"Colunas: {len(consolidado.columns)}")

    return consolidado
```

```python
sinistros = consolidar("sinistros")
pessoas = consolidar("pessoas")
veiculos = consolidar("veiculos")
```
